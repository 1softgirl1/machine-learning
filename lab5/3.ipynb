В файле water.txt представлено 61 наблюдение. Каждое наблюдение – город в
Англии и Уэльсе. Города дополнительно поделены на северные и южные. Для
каждого города известны средняя годовая смертность на 100000 населения (по
данным 1958–1964) и концентрация кальция в питьевой воде (в частях на

2) Постройте 95% доверительные интервалы для средней годовой смертности по
всем южным и северным городам. Отличаются ли границы интервалов?

```python
from statsmodels.stats.weightstats import _tconfint_generic
import numpy as np

south = df[df['location'] == 'Sout']['mortalit']
south_mean = south.mean()
south_std = south.std() / np.sqrt(south.shape[0])
print('95% interval south:', _tconfint_generic(south_mean, south_std, south.shape[0] - 1, 0.005, 'two-sided'))

north = df[df['location'] == 'Nort']['mortalit']
north_mean = north.mean()
north_std = north.std() / np.sqrt(north.shape[0])
print('95% interval north:', _tconfint_generic(north_mean, north_std, north.shape[0] - 1, 0.005, 'two-sided'))

```

Средняя годовая смертность в северных городах статистически значимо выше, чем в южных.

Постройте 95% доверительные интервалы для средней концентрации кальция в
питьевой воде для южных и северных городов. Отличаются ли границы
интервалов?

```python
south = df[df['location'] == 'Sout']['hardness']
south_mean = south.mean()
south_std = south.std() / np.sqrt(south.shape[0])
print('95% interval south:', _tconfint_generic(south_mean, south_std, south.shape[0] - 1, 0.005, 'two-sided'))

north = df[df['location'] == 'Nort']['hardness']
north_mean = north.mean()
north_std = north.std() / np.sqrt(north.shape[0])
print('95% interval north:', _tconfint_generic(north_mean, north_std, north.shape[0] - 1, 0.005, 'two-sided'))
```

Концентрация кальция в питьевой воде статистически значимо выше в южных городах, чем в северных.

В результате анализа установлено, что средняя годовая смертность статистически значимо выше в северных городах, чем в южных. Одновременно с этим концентрация кальция в питьевой воде статистически значимо выше в южных городах. Это позволяет сделать вывод о наличии обратной связи между уровнем смертности и жёсткостью воды: более высокая концентрация кальция ассоциируется с более низкой смертностью.
